```python
import pandas as pd

csv_path = "../02_Cleaned_Data/breast_cancer_cleaned.csv"
df = pd.read_csv(csv_path)

print("Dataset columns:")
for col in df.columns:
    print("-", col)

print("\nFirst 5 records:")
display(df.head())

print("\nData types:")
print(df.dtypes)
```

```python
# Source CSV column -> Warehouse dimension column mapping

dimension_mapping = {
    "Dim_Patient": {
        "Age": "Age",
        "Breast": "Breast",
        "Menopause_ge40": "Menopause_ge40",
        "Menopause_lt40": "Menopause_lt40",
        "Menopause_premeno": "Menopause_premeno"
    },
    "Dim_Tumor": {
        "Tumor-Size": "Tumor_Size",
        "Inv-Nodes": "Inv_Nodes",
        "Node-Caps": "Node_Caps",
        "Deg-Malig": "Deg_Malig"
    },
    "Dim_Clinical": {
        "Irradiat": "Irradiat"
    },
    "Dim_Location": {
        "Breast-Quad_central": "Breast_Quad_central",
        "Breast-Quad_left_low": "Breast_Quad_left_low",
        "Breast-Quad_left_up": "Breast_Quad_left_up",
        "Breast-Quad_right_low": "Breast_Quad_right_low",
        "Breast-Quad_right_up": "Breast_Quad_right_up"
    },
    "Dim_Diagnosis": {
        "Class": "Class"
    }
}

# Validate the source columns
source_columns = set(df.columns)
mapped_columns = {
    col
    for mapping in dimension_mapping.values()
    for col in mapping.keys()
}

print("Source columns:", len(source_columns))
print("Mapped source columns:", len(mapped_columns))
print("Unmapped columns:", source_columns - mapped_columns)
print("Missing source columns:", mapped_columns - source_columns)

if source_columns == mapped_columns:
    print("\nSUCCESS: All source columns are mapped.")
else:
    print("\nCHECK REQUIRED: The mapping does not match the source columns.")
```

```python
# Transform source data into dimension tables

# 1. Patient dimension
dim_patient = df[
    ["Age", "Breast", "Menopause_ge40",
     "Menopause_lt40", "Menopause_premeno"]
].drop_duplicates().reset_index(drop=True)

dim_patient.insert(0, "Patient_Key", range(1, len(dim_patient) + 1))


# 2. Tumor dimension
dim_tumor = df[
    ["Tumor-Size", "Inv-Nodes", "Node-Caps", "Deg-Malig"]
].drop_duplicates().reset_index(drop=True)

dim_tumor = dim_tumor.rename(columns={
    "Tumor-Size": "Tumor_Size",
    "Inv-Nodes": "Inv_Nodes",
    "Node-Caps": "Node_Caps",
    "Deg-Malig": "Deg_Malig"
})

dim_tumor.insert(0, "Tumor_Key", range(1, len(dim_tumor) + 1))


# 3. Clinical dimension
dim_clinical = df[["Irradiat"]].drop_duplicates().reset_index(drop=True)
dim_clinical.insert(0, "Clinical_Key", range(1, len(dim_clinical) + 1))


# 4. Location dimension
location_cols = [
    "Breast-Quad_central",
    "Breast-Quad_left_low",
    "Breast-Quad_left_up",
    "Breast-Quad_right_low",
    "Breast-Quad_right_up"
]

dim_location = df[location_cols].drop_duplicates().reset_index(drop=True)

dim_location = dim_location.rename(columns={
    "Breast-Quad_central": "Breast_Quad_central",
    "Breast-Quad_left_low": "Breast_Quad_left_low",
    "Breast-Quad_left_up": "Breast_Quad_left_up",
    "Breast-Quad_right_low": "Breast_Quad_right_low",
    "Breast-Quad_right_up": "Breast_Quad_right_up"
})

dim_location.insert(0, "Location_Key", range(1, len(dim_location) + 1))


# 5. Diagnosis dimension
dim_diagnosis = df[["Class"]].drop_duplicates().reset_index(drop=True)
dim_diagnosis.insert(0, "Diagnosis_Key", range(1, len(dim_diagnosis) + 1))


print("Dimension tables created successfully!")
print("Dim_Patient:", len(dim_patient), "records")
print("Dim_Tumor:", len(dim_tumor), "records")
print("Dim_Clinical:", len(dim_clinical), "records")
print("Dim_Location:", len(dim_location), "records")
print("Dim_Diagnosis:", len(dim_diagnosis), "records")
```

```python
# Build lookup dictionaries using consistent column names

# 1. Patient lookup
patient_cols = [
    "Age", "Breast", "Menopause_ge40",
    "Menopause_lt40", "Menopause_premeno"
]

patient_lookup = {
    tuple(row[c] for c in patient_cols): row["Patient_Key"]
    for _, row in dim_patient.iterrows()
}

# 2. Tumor lookup: use original source columns
tumor_cols = ["Tumor-Size", "Inv-Nodes", "Node-Caps", "Deg-Malig"]

tumor_lookup = {
    tuple(row[c] for c in tumor_cols): key
    for (_, row), key in zip(
        df[tumor_cols].drop_duplicates().iterrows(),
        dim_tumor["Tumor_Key"]
    )
}

# 3. Clinical lookup
clinical_lookup = dict(
    zip(dim_clinical["Irradiat"], dim_clinical["Clinical_Key"])
)

# 4. Location lookup: source names -> warehouse names
location_source_cols = [
    "Breast-Quad_central",
    "Breast-Quad_left_low",
    "Breast-Quad_left_up",
    "Breast-Quad_right_low",
    "Breast-Quad_right_up"
]

location_lookup = {
    tuple(row[c] for c in location_source_cols): key
    for (_, row), key in zip(
        df[location_source_cols].drop_duplicates().iterrows(),
        dim_location["Location_Key"]
    )
}

# 5. Diagnosis lookup
diagnosis_lookup = dict(
    zip(dim_diagnosis["Class"], dim_diagnosis["Diagnosis_Key"])
)

# Build fact table
fact = pd.DataFrame({
    "Fact_Key": range(1, len(df) + 1),

    "Patient_Key": [
        patient_lookup[tuple(row[c] for c in patient_cols)]
        for _, row in df.iterrows()
    ],

    "Tumor_Key": [
        tumor_lookup[tuple(row[c] for c in tumor_cols)]
        for _, row in df.iterrows()
    ],

    "Clinical_Key": df["Irradiat"].map(clinical_lookup),

    "Location_Key": [
        location_lookup[tuple(row[c] for c in location_source_cols)]
        for _, row in df.iterrows()
    ],

    "Diagnosis_Key": df["Class"].map(diagnosis_lookup),

    "Patient_Count": 1
})

print("Fact table created successfully!")
print("Fact table shape:", fact.shape)
print("\nMissing foreign keys:")
print(fact.isnull().sum())

display(fact.head())
```

```python
# Validate fact table foreign-key relationships

checks = {
    "Patient_Key": set(dim_patient["Patient_Key"]),
    "Tumor_Key": set(dim_tumor["Tumor_Key"]),
    "Clinical_Key": set(dim_clinical["Clinical_Key"]),
    "Location_Key": set(dim_location["Location_Key"]),
    "Diagnosis_Key": set(dim_diagnosis["Diagnosis_Key"])
}

foreign_key_columns = {
    "Patient_Key": "Patient_Key",
    "Tumor_Key": "Tumor_Key",
    "Clinical_Key": "Clinical_Key",
    "Location_Key": "Location_Key",
    "Diagnosis_Key": "Diagnosis_Key"
}

all_valid = True

for fact_col, dimension_col in foreign_key_columns.items():
    invalid = set(fact[fact_col].dropna()) - checks[dimension_col]
    print(f"{fact_col}: {len(invalid)} invalid keys")

    if invalid:
        all_valid = False

print("\nAll foreign keys valid:", all_valid)
print("Fact records:", len(fact))
print("Expected records:", len(df))
print("Record count matches:", len(fact) == len(df))
```

```python
import sqlite3

db_path = "../04_Data_Warehouse/breast_cancer_warehouse.db"

# Confirm the tables are still empty before loading
conn = sqlite3.connect(db_path)

try:
    cursor = conn.cursor()

    tables = [
        "Dim_Patient",
        "Dim_Tumor",
        "Dim_Clinical",
        "Dim_Location",
        "Dim_Diagnosis",
        "Fact_Breast_Cancer"
    ]

    for table in tables:
        cursor.execute(f"SELECT COUNT(*) FROM {table}")
        if cursor.fetchone()[0] != 0:
            raise RuntimeError(
                f"{table} is no longer empty. Loading cancelled."
            )

    # Enable foreign-key checking on this connection
    cursor.execute("PRAGMA foreign_keys = ON")

    # Load dimension tables first
    dim_patient.to_sql("Dim_Patient", conn, if_exists="append", index=False)
    dim_tumor.to_sql("Dim_Tumor", conn, if_exists="append", index=False)
    dim_clinical.to_sql("Dim_Clinical", conn, if_exists="append", index=False)
    dim_location.to_sql("Dim_Location", conn, if_exists="append", index=False)
    dim_diagnosis.to_sql("Dim_Diagnosis", conn, if_exists="append", index=False)

    # Load the fact table after its dimensions
    fact.to_sql("Fact_Breast_Cancer", conn, if_exists="append", index=False)

    # Verify foreign-key constraints
    cursor.execute("PRAGMA foreign_key_check")
    violations = cursor.fetchall()

    if violations:
        raise RuntimeError(f"Foreign-key violations found: {violations}")

    conn.commit()
    print("ETL LOAD SUCCESSFUL!")

except Exception as e:
    conn.rollback()
    print("ETL LOAD FAILED — transaction rolled back.")
    print("Error:", e)

finally:
    conn.close()
```
